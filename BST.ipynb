```python
class Node:

    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
```

```python
class BST:

    def __init__(self):
        self.root = None
```

```python
def insert(self, root, key):

    # Empty position found
    if root is None:
        return Node(key)

    # Go to left subtree
    if key < root.key:
        root.left = self.insert(root.left, key)

    # Go to right subtree
    elif key > root.key:
        root.right = self.insert(root.right, key)

    # Duplicate values are ignored
    return root
```
